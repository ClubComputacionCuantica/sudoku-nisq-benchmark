## Stage 6-7: Metrics (Evaluation + Normalization)

**Purpose**: Auto-compute and store benchmark metrics

**Storage**: `{puzzle_hash}/stage_6_7_metrics.json`

**Stage 6 (Evaluation - σ)**:
- Success probability p_succ with 95% CI
- Distinct valid solutions (coverage)
- Signal-to-noise ratio (SNR)
- Ranking metrics (top-k, precision@k, recall@k)

**Stage 7 (Normalization - τ)**:
- Gate efficiency: η_gate = p_succ / G_{2q}
- Volume efficiency: η_volume = p_succ / (w × d_c)
- Shot efficiency: η_shot = p_succ / shots

**Status**: 🔲 Placeholder (Phase 4 implementation)

```python
# Initialize Stage 6-7 manager
stage67 = MetricsMetadataManager(cache_base, puzzle_hash)
print(f"Stage 6-7 manager: {stage67}")
print(f"Combines evaluation (Stage 6) and normalization (Stage 7)")

# Mock validator
valid_bitstrings = {"0000111100001111", "1111000011110000"}
validator = lambda bs: bs in valid_bitstrings

# Mock circuit resources (from Stage 3)
circuit_resources = {
    "n_qubits": 16,
    "depth": 203,
    "cx": 128,  # Two-qubit gate count
}

# Try to record metrics (will fail - placeholder)
try:
    stage67.record(
        run_id=run_id,
        counts=mock_counts,
        validator=validator,
        circuit_resources=circuit_resources,
        shots=1024,
    )
    print(f"✅ Metrics computed and recorded for {run_id}")
except NotImplementedError as e:
    print(f"❌ Expected: {e}")
    
    # Show what WOULD be computed
    print("\n🔧 What would be computed:")
    
    # Stage 6: Evaluation
    total_shots = sum(mock_counts.values())
    valid_shots = sum(c for bs, c in mock_counts.items() if validator(bs))
    p_succ = valid_shots / total_shots
    distinct = len([bs for bs in mock_counts if validator(bs)])
    snr = p_succ / (1 - p_succ) if p_succ < 1.0 else float('inf')
    
    print(f"  Stage 6 (Evaluation):")
    print(f"    p_succ: {p_succ:.4f}")
    print(f"    distinct_solutions: {distinct}")
    print(f"    snr: {snr:.2f}")
    
    # Stage 7: Normalization
    eta_gate = p_succ / circuit_resources["cx"]
    eta_volume = p_succ / (circuit_resources["n_qubits"] * circuit_resources["depth"])
    eta_shot = p_succ / 1024
    
    print(f"  Stage 7 (Normalization):")
    print(f"    η_gate: {eta_gate:.6f}")
    print(f"    η_volume: {eta_volume:.6f}")
    print(f"    η_shot: {eta_shot:.6f}")
```
